# Hospital Analytics — Exploratory Data Analysis

Reads the hostable sample produced by `python -m src.etl.pipeline` (`data/sample/hospital_cleaned_sample.csv`). Columns are PascalCase to match SQL, Excel, Power BI, and Streamlit.

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sample = root / "data" / "sample" / "hospital_cleaned_sample.csv"
df = pd.read_csv(sample, parse_dates=["DateOfAdmission", "DischargeDate"])
df.head()
```

```python
df.info()
df.describe(include="all")
```

## Revenue by Department

```python
dept_revenue = df.groupby("Department")["BillingAmount"].sum().sort_values(ascending=False)
dept_revenue.plot(kind="bar", figsize=(10, 5), title="Total Revenue by Department")
plt.ylabel("Revenue ($)")
plt.tight_layout()
plt.show()
```

## Monthly Admissions Trend

```python
monthly = df.groupby(["AdmissionYear", "AdmissionMonth"]).size()
monthly.plot(kind="line", figsize=(12, 5), title="Monthly Admissions Trend", marker="o")
plt.ylabel("Admissions")
plt.tight_layout()
plt.show()
```

## Top Doctors by Patient Count

```python
top_doctors = df["Doctor"].value_counts().head(10)
top_doctors.plot(kind="barh", figsize=(8, 6), title="Top 10 Doctors by Patient Count")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Age Group and Gender

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df["AgeGroup"].value_counts().plot(kind="bar", ax=axes[0], title="Age Group Distribution")
df["Gender"].value_counts().plot(kind="pie", ax=axes[1], autopct="%1.1f%%", title="Gender Distribution")
plt.tight_layout()
plt.show()
```

## Insurance Provider Breakdown

```python
df["InsuranceProvider"].value_counts().plot(kind="bar", figsize=(8, 5), title="Patients by Insurance Provider")
plt.tight_layout()
plt.show()
```
